## Situation 3 - Content-based filtering  *(medium)*

**Studio problem.** Popularity gives everyone the same list. Content-based
filtering personalises with **item features**. First build **Phuc's taste
profile**: a vector of 4 numbers, one per feature (`coffee`, `sweet`, `iced`, `tea`).

```
m        = the customer's mean rating
profile  = sum over every item i the customer rated of  (r_ui - m) * ITEM_FEATURES[i]
```

Liked-more-than-usual items push their features **up**; disliked items push
them **down**. Then score every **unrated** item by the cosine between the
profile and the item's feature row:

```
cosine(a, b) = a . b / (|a| |b|)          (return 0.0 if either vector is all zeros)
```

Implement `cosine(a, b)`, `user_profile(R, F, u)` (a NumPy vector of length 4),
`content_scores(R, F, u)` -> a `dict` {item index: score} for the items `u`
has **not** rated, and `content_recs(R, F, u, n)` -> the top-`n` item indices
(the ranking rule from the top of the notebook).

Then store:

- `S3_profile` - Phuc's profile as a list of 4 floats, each rounded to 4
- `S3_liked_feature` - the **name** of the feature with the largest profile value
- `S3_top3` - the 3 recommended drink **names**; `S3_top_score` - the best score (4 decimals)
- `S3_comment` - **2-3 sentences**: which features does *your* Phuc like and
  dislike, and how do they explain the top-3? Name one drink content-based
  can never recommend to Phuc and say why (the *filter bubble*).

````python
def cosine(a, b):
    a = np.asarray(a, dtype=float); b = np.asarray(b, dtype=float)
    # TODO: YOUR CODE HERE -> a @ b / (norm(a) * norm(b)), or 0.0 if a norm is 0
    na, nb = np.linalg.norm(a), np.linalg.norm(b)
    if na == 0 or nb == 0:
        return 0.0
    return float((a @ b) / (na * nb))

def user_profile(R, F, u):
    # TODO: YOUR CODE HERE -> sum of (r_ui - mean_u) * F[i] over the items u rated
    rated = ~np.isnan(R[u])
    mean_u = np.nanmean(R[u])
    return np.sum((R[u, rated] - mean_u)[:, None] * F[rated], axis=0)

def content_scores(R, F, u):
    p = user_profile(R, F, u)
    # TODO: YOUR CODE HERE -> {i: cosine(p, F[i]) for every item i that u has NOT rated}
    return {i: cosine(p, F[i]) for i in range(R.shape[1]) if np.isnan(R[u, i])}

def content_recs(R, F, u, n):
    sc = content_scores(R, F, u)
    # TODO: YOUR CODE HERE -> top-n indices: sorted(sc, key=lambda i: (-sc[i], i))[:n]
    return sorted(sc, key=lambda i: (-sc[i], i))[:n]

S3_profile = [round(float(x), 4) for x in user_profile(R, ITEM_FEATURES, TARGET)]         # TODO: YOUR CODE HERE -> [round(float(x), 4) for x in user_profile(R, ITEM_FEATURES, TARGET)]
S3_liked_feature = FEATURES[int(np.argmax(user_profile(R, ITEM_FEATURES, TARGET)))]   # TODO: YOUR CODE HERE
S3_top3 = [ITEMS[i] for i in content_recs(R, ITEM_FEATURES, TARGET, TOP_N)]            # TODO: YOUR CODE HERE -> names
S3_top_score = round(float(content_scores(R, ITEM_FEATURES, TARGET)[content_recs(R, ITEM_FEATURES, TARGET, TOP_N)[0]]), 4)       # TODO: YOUR CODE HERE
S3_comment = "Phuc strongly likes the 'tea' feature (+1.2) because he gave Hot Ginger Tea 5 stars, while he dislikes 'sweet' (-1.6), 'coffee' (-1.2), and 'iced' (-1.2) after rating Egg Coffee, Salt Coffee, Coconut Coffee, and Peach Iced Tea at 3 stars. This explains why Lotus Tea tops his list (score 0.5145), followed by Matcha Latte and Black Coffee. Because of the filter bubble, content-based filtering will never recommend Iced Milk Coffee (score -1.0) to Phuc since it combines coffee, sweet, and iced features with zero tea."           # TODO: YOUR CODE HERE -> 2-3 sentences

print("profile", dict(zip(FEATURES, S3_profile or [])), "| likes most:", S3_liked_feature)
print("content top-3 for", TARGET_USER, ":", S3_top3, "| best score", S3_top_score)
````

## Situation 5 - Predict and recommend  *(hard)*

**Studio problem.** Now fill a blank: how many stars would **Phuc** give
**Iced Milk Coffee**? Use Phuc's neighbours who **did** rate it. Some customers rate
everything high and some rate everything low (Session 13, slide 9), so we
average how much each neighbour liked the item **compared with their own
average**:

```
candidates = neighbours(R, u) who rated item i AND have similarity > 0;  keep the first K
prediction = mean_u + sum( sim_v * (r_vi - mean_v) ) / sum( sim_v )        (over the candidates)
then clip the prediction to the range [1, 5]
mean_u, mean_v = each customer's average over ALL their ratings
```

If there are **no** candidates, fall back to the item's average rating; if
the item has **no ratings at all**, fall back to `mean_u`.

Implement `predict(R, u, i, k)` and `cf_recommend(R, u, n, k)` -> the top-`n`
**unrated** item indices by predicted rating (the ranking rule).

Then store:

- `S5_focus_pred` - `round(predict(R, TARGET, FOCUS, K), 4)` - Phuc x Iced Milk Coffee
- `S5_top3` - the 3 collaborative-filtering recommendations (**names**) for Phuc
- `S5_top_pred` - the predicted rating of the first one (4 decimals)
- `S5_comment` - **3-4 sentences**: put *your* three top-3 lists side by side
  (popularity S2, content S3, collaborative S5). Which drinks appear in more

````python
def predict(R, u, i, k):
    cand = [(v, s) for v, s in neighbours(R, u) if not np.isnan(R[v, i]) and s > 0][:k]   # TODO: YOUR CODE HERE -> [(v, s) for v, s in neighbours(R, u) if R[v, i] is rated and s > 0][:k]
    # TODO: YOUR CODE HERE -> if cand: mean_u + sum(s * (R[v, i] - mean_v)) / sum(s), clipped to [1, 5]
    mean_u = float(np.nanmean(R[u]))
    if cand:
        num = sum(s * (R[v, i] - float(np.nanmean(R[v]))) for v, s in cand)
        den = sum(s for _, s in cand)
        return float(np.clip(mean_u + num / den, 1.0, 5.0))
    # TODO: YOUR CODE HERE -> otherwise the item's mean rating, or mean_u if nobody rated the item
    col = R[:, i]
    if np.any(~np.isnan(col)):
        return float(np.nanmean(col))
    return mean_u

def cf_recommend(R, u, n, k):
    # TODO: YOUR CODE HERE -> predict every item u has NOT rated; top-n indices by (-prediction, index)
    unrated = [i for i in range(R.shape[1]) if np.isnan(R[u, i])]
    preds = {i: predict(R, u, i, k) for i in unrated}
    return sorted(unrated, key=lambda i: (-preds[i], i))[:n]

S5_focus_pred = round(predict(R, TARGET, FOCUS, K), 4)   # TODO: YOUR CODE HERE
S5_top3 = [ITEMS[i] for i in cf_recommend(R, TARGET, TOP_N, K)]         # TODO: YOUR CODE HERE -> names
S5_top_pred = round(predict(R, TARGET, cf_recommend(R, TARGET, TOP_N, K)[0], K), 4)     # TODO: YOUR CODE HERE
S5_comment = "Comparing our three top-3 lists for Phuc—Popularity (Black Coffee, Iced Milk Coffee, Cold Brew), Content-based (Lotus Tea, Matcha Latte, Black Coffee), and Collaborative Filtering (Black Coffee, Lotus Tea, Cold Brew)—Black Coffee appears in all three lists while Lotus Tea and Cold Brew each appear in two. It is surprising that Collaborative Filtering recommends Cold Brew (3.6522) over Matcha Latte because Binh (whose raw cosine similarity is high despite opposite tea preferences) pulls Matcha Latte's prediction down. I would show Phuc the Content-based list (or a hybrid with Black Coffee and Lotus Tea) because it best captures his love for tea and non-sweet drinks."        # TODO: YOUR CODE HERE -> 3-4 sentences

print("%s x %s -> %s stars" % (TARGET_USER, FOCUS_ITEM, S5_focus_pred))
print("popularity   :", S2_top3)
print("content      :", S3_top3)
print("collaborative:", S5_top3, "| best predicted", S5_top_pred)
````
